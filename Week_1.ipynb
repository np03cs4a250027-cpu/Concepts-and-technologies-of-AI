```python
#Prove A.A−1 = I.

A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

A_inv = np.linalg.inv(A)
res = np.round(np.dot(A, A_inv))
print("A.A−1:",res)
print("Identity matrix(I) : ",np.eye(2))
```

```python
#Prove AB ̸= BA.

AB=np.dot(A,B)
BA=np.dot(B,A)
print("AB: ",AB)
print("BA: ",BA)
print("AB and BA is not equal")
```

```python
#Prove (AB)T = BTAT.

A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

A_transpose =A.T
B_transpose =B.T
AB_transpose = np.dot(A,B).T
tran=np.dot(B_transpose,A_transpose)
print("AB Transpos: e",AB_transpose )
print("B Transpose dot A_transpose: ",tran)
```

```python
#Solve the following system of Linear equation using Inverse Methods.

#2x − 3y + z = −1
#x − y + 2z = −3
#3x + y − z = 9

A = np.array([[2, -3, 1],[1, -1, 2],[3, 1, -1]])
B = np.array([-1, -3, 9])
A_inv = np.linalg.inv(A)
X = np.dot(A_inv, B)
print("Solution [x, y, z]:", np.round(X))
```

```python

#Element-wise Addition:

import time
import numpy as np

size = 1000000

list1 = list(range(size))
list2 = list(range(size))


start = time.time()
result_list = [list1[i] + list2[i] for i in range(size)]
end = time.time()

print("Python List Addition Time:", end - start, "seconds")


arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
result_numpy = arr1 + arr2
end = time.time()
print("NumPy Addition Time:", end - start, "seconds")
```

```python
#Element-wise Multiplication:

import time
import numpy as np

size = 1000000

list1 = list(range(size))
list2 = list(range(size))


start = time.time()
result_list = [list1[i] * list2[i] for i in range(size)]
end = time.time()

print("Python List Multiplication Time:", end - start, "seconds")


arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
result_numpy = arr1 * arr2
end = time.time()
print("NumPy Multiplication Time:", end - start, "seconds")
```

```python

#Dot Product:

import time
import numpy as np

size = 1000000

list1 = list(range(size))
list2 = list(range(size))


start = time.time()
result_list = sum(list1[i] * list2[i] for i in range(size))
end = time.time()
print("Python List Dot Product Time:", end - start, "seconds")


arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
result_numpy = np.dot(arr1, arr2)
end = time.time()
print("NumPy Dot Product Time:", end - start, "seconds")
```

```python
#Matrix Multiplication


import time
import numpy as np

n = 1000

matrix1 = [[1 for j in range(n)] for i in range(n)]
matrix2 = [[2 for j in range(n)] for i in range(n)]


start = time.time()
result_list = [
    [sum(matrix1[i][k] * matrix2[k][j] for k in range(n))for j in range(n)]
    for i in range(n)]
end = time.time()
print("Python List Matrix Multiplication Time:", end - start, "seconds")


arr1 = np.array(matrix1)
arr2 = np.array(matrix2)
start = time.time()
result_numpy = np.dot(arr1, arr2)
end = time.time()
print("NumPy Matrix Multiplication Time:", end - start, "seconds")
```
