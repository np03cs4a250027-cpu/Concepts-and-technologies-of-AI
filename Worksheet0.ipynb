```python
# 1. List of study time classifications.

time_data = [
(3.5, 2.0, 7.0), (5.0, 1.5, 6.5), (2.5, 3.0, 8.0),
(4.0, 2.0, 6.0), (1.5, 4.5, 9.0), (3.0, 2.5, 7.5),
(5.5, 1.0, 6.0), (2.0, 3.5, 8.5), (4.5, 2.0, 7.0),
(3.0, 3.0, 7.5), (6.0, 1.5, 6.0), (2.5, 4.0, 8.0),
(4.0, 2.5, 7.0), (5.0, 2.0, 6.5), (3.5, 2.5, 7.0)
]

low_study = []
moderate_study = []
high_study = []

for day in time_data:
    if day[0] < 3:
        low_study.append(day[0])
    elif day[0] <= 5:
        moderate_study.append(day[0])
    else:
        high_study.append(day[0])

print("Low study time:", low_study)
print("Moderate study time:", moderate_study)
print("High study time:", high_study)
```

```python
#2.  Number of days

print("Low study time",len(low_study))
print("Moderate study time",len(moderate_study))
print("High study time",len(high_study))
```

```python
#3.  Convert Study Hours to Minutes:

converted_values=[]
for day in time_data:
        converted_values.append(day[0] * 60)
print("The hours conveted in minutes",converted_values)


```

```python
#4. Analyze Average Time Use:

study_hours=[]
entertainment_hours=[] 
sleep_hours=[]
for day in time_data:
    study_hours.append(day[0])
    entertainment_hours.append(day[1])
    sleep_hours.append(day[2])
avg_study= sum(study_hours)/len(study_hours)
avg_entertainment=sum(entertainment_hours)/len(entertainment_hours)
avg_sleep = sum(sleep_hours)/len(sleep_hours)

print("Average study hours:",avg_study)
print("Average entertainment hours:",avg_entertainment)
print("Average sleep hours:",avg_sleep)



```

```python
import matplotlib.pyplot as plt
    
plt.scatter(study_hours, sleep_hours, color='blue')
plt.xlabel("Study Hours")
plt.ylabel("Sleep Hours")
plt.title("Relationship Between Study Hours and Sleep Hours")
plt.show()
```

```python
# Task 1 - Sum of Nested Lists:

def sum_nested_list(nested_list):
    total=0
    for item in nested_list:
        if isinstance(item,list):
            total+= sum_nested_list(item)
        else:
            total+=item
    return total
nested_list = [1, [2, [3, 4], 5], 6, [7, 8]]
result = sum_nested_list(nested_list)
print("The total sum is ", result)
```

```python

def generate_permutations(s):
    if len(s) == 0:
        return [""]

    permutations = []

    for i in range(len(s)):
        char = s[i]
        remaining = s[:i] + s[i+1:]

        for p in generate_permutations(remaining):
            permutations.append(char + p)

    return list(set(permutations))


print(generate_permutations("abc"))
print(generate_permutations("aab"))
```

```python
def calculate_directory_size(directory):
    total = 0

    for value in directory.values():
        if isinstance(value, dict):
            total += calculate_directory_size(value)
        else:
            total += value

    return total


directory_structure = {
    "file1.txt": 200,
    "file2.txt": 300,
    "subdir1": {
        "file3.txt": 400,
        "file4.txt": 100
    },
    "subdir2": {
        "subsubdir1": {
            "file5.txt": 250
        },
        "file6.txt": 150
    }
}

result = calculate_directory_size(directory_structure)

print("Total directory size:", result, "KB")
```

```python
def min_coins(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1
    
print("Minimum coins",min_coins([1, 2, 5], 11))
print("Minimum coins",min_coins([2],3))


```

```python
def longest_common_subsequence(s1, s2):
    m = len(s1)
    n = len(s2)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[m][n]


print("The longest common subsequence is",longest_common_subsequence("abcde", "ace"))
```

```python

def knapsack(weights, values, capacity):
    n = len(weights)
    dp = [[0 for x in range(capacity + 1)] for y in range(n + 1)]
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(values[i - 1] + dp[i - 1][w - weights[i - 1]],dp[i - 1][w])
            else:
                dp[i][w] = dp[i - 1][w]
    return dp[n][capacity]

weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7
result = knapsack(weights, values, capacity)
print("Maximum value:", result)
```
